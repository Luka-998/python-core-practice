## **1) Object Representation**

```python
# __str__ -> return string , for user
# __repr__ -> method defines the official string representation of an object, primarily used for debugging. 
# By default, it shows the type and memory address of the object
```

```python
class H:
    def __init__(self,val):
        self.val=val

    def __str__(self):
        return str(self.val)
    def __repr__(self):
        return f"This val: {self.val}"
```

```python
human = H(14)
print(human)
human
```

```python
class Plus:
    def __init__(self,x,y):
        self.x = x
        self.y = y
    def __add__(self,z):
        self.current = self.x+self.y
        self.new = self.current+z
        if self.new < 30:
            self.new = self.current + z
        return self.new
```

MetricTracker gets input such as model_name:str; metrics:dict

```python
model_name = 'LogisticRegression'
metrics = {'f1':0.6,'recall':0.55,'accuracy':0.83}
```

```python
class MetricTracker:
    def __init__(self,name,metrics):
        self.name = name
        self.metrics = metrics
    def __repr__(self):

        return (
            f"Model: {self.name}",
            f"{self.metrics}:{[str(metric) for metric in self.metrics.keys()]}",
        )
```

### Task2: ___repr___ and ___add___ 



Class: ExperimentResult
constructor takes in [model name:str, metrics: dict[str,float]]

__ add __ should return new object exp1+exp2 ( docs: " object.__add__(self, other)")

```python
class ExperimentResult:
    def __init__(self,model_name,metrics):
        self.name = model_name
        self.metrics = metrics 
    def __repr__(self):
        return f"Class name: {self.__class__.__name__}, Model name: {self.name}, metrics: {self.metrics}"   
    def __add__(self,other):
        if isinstance(other,self.__class__):
            if other.name != self.name:
                raise ValueError("Not the same model!")
            if other.metrics.keys()!=self.metrics.keys():
                raise ValueError("Not the same metrics!")
            else:
                # make average metrics
                new_metrics = {}
                for k,v in other.metrics.items():
                    new_metrics[k] = (self.metrics[k] + v)/2
                result = ExperimentResult(self.name,new_metrics)
                return result
        else:
                return NotImplemented
         

exp1 = ExperimentResult('DT',{'accuracy':0.96,'recall':0.33})
exp2 = ExperimentResult('RandomForest',{'accuracy':0.62,'recall':0.73})
exp3 = ExperimentResult('DT',{'accuracy':0.86,'recall':0.43})
e = exp1 + exp3
print(e==exp1)
print(e==exp3)
print(id(exp1),id(exp3),id(e))
```
