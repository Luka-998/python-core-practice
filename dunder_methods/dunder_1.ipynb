## **1) Object Representation**

```python
# __str__ -> return string , for user
# __repr__ -> method defines the official string representation of an object, primarily used for debugging. 
# By default, it shows the type and memory address of the object
```

```python
class H:
    def __init__(self,val):
        self.val=val

    def __str__(self):
        return str(self.val)
    def __repr__(self):
        return f"This val: {self.val}"
```

```python
human = H(14)
print(human)
human
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np
```

```python
lg = LinearRegression(fit_intercept=True)
X = np.array([[0,0,2],[0,0,1]])
y = np.array([0,0,1])
```

```python
class Model:
    def __init__(self,mod,X,y):
        self.mod = mod
        self.x_train = X
        self.y_train = y

    def __repr__(self):
        return (
            f"X Shape: {self.x_train.shape[0]} rows\n"
            f"X shape: {self.x_train.shape[1]} cols"
        )
```

```python
l = Model(lg,X,y)
print(repr(l))
```

```python
class Plus:
    def __init__(self,x,y):
        self.x = x
        self.y = y
    def __add__(self,z):
        self.current = self.x+self.y
        self.new = self.current+z
        if self.new < 30:
            self.new = self.current + z
        return self.new
```

```python
calculator = Plus(14,19)
calculator.__add__(90)
```

MetricTracker gets input such as model_name:str; metrics:dict

```python
model_name = 'LogisticRegression'
metrics = {'f1':0.6,'recall':0.55,'accuracy':0.83}
```

```python
class MetricTracker:
    def __init__(self,name,metrics):
        self.name = name
        self.metrics = metrics
    def __repr__(self):

        return (
            f"Model: {self.name}",
            f"{self.metrics}:{[str(metric) for metric in self.metrics.keys()]}",
        )
```
