### Python scopes and namespaces

A namespace is a mapping from names to objects.

- Examples of namespaces are: the set of built-in names (containing functions such as abs(), and built-in exception names); 

- the global names in a module; and the local names in a function invocation. In a sense the set of attributes of an object also form a namespace. 

- The important thing to know about namespaces is that there is absolutely no relation between names in different namespaces; for instance, two different modules may both define a function maximize without confusion — users of the modules must prefix it with the module name.

### The statements executed by the top-level invocation of the interpreter, either read from a script file or interactively, are considered part of a module called __main__, so they have their own global namespace.

### Local namespace is created at the runtime, when the function is called and delleted when function returns or raises the exception.


### **The scopes are used dynamically: (at time of execution there are 3 or 4 nested scopes whose namespace are accessible directly)**

#### 1) Innermost scope, searched first, contains local names

#### 2) the scope of any enclosing functions, which are searched starting with the nearest enclosing scope, contain non-local but also non-global names

#### 3) the next-to-last scope contains the current module’s global names

#### 4) the outermost scope (searched last) is the namespace containing built-in names

LEGB RULE

1) LOCAL

2) ENCLOSING

3) GLOBAL

4) Built-in

```python
class Scope:
    # No Class attributes!
    def __init__(self,value,name):
        self.value = value # instance attributes 
        self.name = name 
    def do_something(self,func):

        print(type(func)) 
        print("Local namespace")

        def wrapper(*args,**kwargs): # inner function
            print(f"Enclosing namespace: {self.name}\n{self.value}") # SELF comes from enclosing scope of do_something method / self.name +self.value are found as attribtes of the object that self references
            result = func(*args,**kwargs) # inside the wrapper func references original function thanks to closure
            return result
        return wrapper

c = Scope(15,"Luka")

def lower_string(text):
    return str(text).lower()

func = c.do_something(lower_string) # reference to the inner function!
result_lower_case = func("LUKA JASOVIC UPPERCASE")
print(result_lower_case)
```
