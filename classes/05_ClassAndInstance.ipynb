### Class objects supports: 

- Instantiation

- Attribute reference (Class attributes vs Instance attributes)

```python
class Dog:
    favorite_food = "Bones"
    def __init__(self,size,agressive):
        self.s = size
        self.agr = agressive

    def instance(self):
        return f"Return only instance attributes: {self.s} and {self.agr}"

    def shared(self):
        return f"Return only shared attribute: {favorite_food}" # Can't be called like this! It's called directly from the Class instance obj


dog1 = Dog('small','Yes agressive')
print(dog1.instance())
dog2 = Dog('big','Not agressive')
print(dog2.instance())

print(dog1.favorite_food == dog2.favorite_food) #Shared  attribute, same between the instance
```

### Random Remarks:

#### If the same attribute name occurs in both an instance and in a class, then attribute lookup prioritizes the instance:




```python
class Warehouse:
    purpose = "storage"
    region = "west"


S3 = Warehouse()

print(S3.purpose)
print(S3.region)
```

```python
E2 = Warehouse()
E2.region = "EAST"

print(E2.purpose)
print(E2.region) # EAST is expected here
```

### Function defined outside of the class object can be used within the class without any problems


```python
def get_sum(self,x,y):
    return x + y

class One:
    z = get_sum
    def __init__(self,name,numbers):
        self.name = name
        self.numbers = numbers

    def g(self):
        return "hello world"

    h = g


onne = One('Luka',10)
print(onne.z)
print(onne.z(1,2))
```

### Methods can be called by another methods as well:

```python
class Build:
    storage = []
    def __init__(self,buildname):
        self.name = buildname
    def add(self,value):
        self.storage.append(value)

    def add_twice(self,value):
        self.add(value)
        self.add(value)
```

```python
zzz = Build('Luka')
print(zzz.storage)
zzz.add(5)
print(zzz.storage)
print(zzz.add_twice(10))
print(zzz.storage)
```
